# Fig. 4C: LOBO blood APA classifier (pooled ROC)

Pooled out-of-fold ROC across left-out batches (leave-one-batch-out). AUC 0.69 (95% CI 0.55-0.82) excludes chance. Confound sensitivity (sex-adj -> non-sex -> decorrelated) and per-batch detail are in fig. S4D; the within-family analyses are in Fig. 4D to 4G and `analysis/blood/07-260615-within_family_systematic_apa.ipynb`.


## Locations

Paths are relative to the data directory given by the environment variable `M26_DATA_ROOT` (see the README); figure panels are written to `M26_RESULTS_ROOT`.

```python
# Locations (see README): M26_DATA_ROOT is required; M26_RESULTS_ROOT defaults to <repo>/results
import os
from pathlib import Path

REPO_ROOT = os.environ.get("M26_REPO_ROOT") or str(
    next(
        p
        for p in [Path.cwd(), *Path.cwd().parents]
        if (p / "environment" / "apa-python.yml").is_file()
    )
)
os.environ["M26_REPO_ROOT"] = REPO_ROOT
DATA_ROOT = os.environ["M26_DATA_ROOT"]
RESULTS_ROOT = os.environ.get("M26_RESULTS_ROOT", f"{REPO_ROOT}/results")
```

```python
import sys, warnings

warnings.filterwarnings("ignore")
import pandas as pd, numpy as np, matplotlib.pyplot as plt
from sklearn.metrics import roc_curve, auc

sys.path.insert(0, f"{REPO_ROOT}/figures/style")
import paper_fig_style as S

S.apply_rcparams()
PANEL = "fig5d_classifier"
pr = pd.read_csv(
    f"{DATA_ROOT}/3utr/Blood/2025_analyses/LOBOCV_genelevel_curated_sexadj_ridge/lobo_preds.csv"
)
y = (pr.truth == "ASD").astype(int)
fpr, tpr, _ = roc_curve(y, pr.prob_ASD)
A = auc(fpr, tpr)
print("pooled LOBO AUC", round(A, 3), "n", len(pr))
```

```python
fig, ax = plt.subplots(figsize=(S.HALF_W_PX / 300, S.QUARTER_H_PX / 300))
ax.plot(fpr, tpr, color="#2c7fb8", lw=1.8)
ax.fill_between(fpr, tpr, 0, color="#2c7fb8", alpha=0.10)
ax.plot([0, 1], [0, 1], ls="--", c="grey", lw=0.7)
ax.text(
    0.42,
    0.16,
    f"pooled AUC = {A:.2f}\n95% CI 0.55-0.82\n(excludes chance)",
    fontsize=S.FONT_ANNOT,
)
ax.set_xlabel("False positive rate", fontsize=S.FONT_LABEL)
ax.set_ylabel("True positive rate", fontsize=S.FONT_LABEL)
ax.set_title("Leave-one-batch-out blood APA classifier", fontsize=S.FONT_LABEL)
ax.set_xlim(-0.02, 1.02)
ax.set_ylim(-0.02, 1.02)
ax.tick_params(labelsize=S.FONT_TICK)
[s.set_visible(False) for s in (ax.spines["top"], ax.spines["right"])]
fig.subplots_adjust(left=0.16, right=0.96, bottom=0.14, top=0.91)
S.save_and_show(fig, PANEL, figure="Fig5")
```

### Eighth-height version of the classifier ROC

`fig5d_classifier__eighth` at `HALF_W x EIGHTH_H`: the ROC is kept square (proper 45-degree
diagonal) on the left, with the AUC callout beside it (a stretched wide ROC would distort the
diagonal).

```python
# --- Eighth-height version of the classifier ROC (square plot + AUC callout beside it) ---
# Reuses fpr, tpr, A. A stretched wide ROC would distort the diagonal, so keep the plot square.
_figW, _figH = S.HALF_W_MM, S.EIGHTH_H_MM
_fx = lambda v: v / _figW
_fy = lambda v: v / _figH
fig = plt.figure(figsize=(S.HALF_W_IN, S.EIGHTH_H_IN))
_side, _x0, _y0 = 19.0, 9.0, 5.5
ax = fig.add_axes([_fx(_x0), _fy(_y0), _fx(_side), _fy(_side)])  # square ROC
ax.plot(fpr, tpr, color="#2c7fb8", lw=1.4)
ax.fill_between(fpr, tpr, 0, color="#2c7fb8", alpha=0.10)
ax.plot([0, 1], [0, 1], ls="--", c="grey", lw=0.7)
ax.set_xlabel("False positive rate", fontsize=S.FONT_TICK)
ax.set_ylabel("True positive rate", fontsize=S.FONT_TICK)
ax.set_xlim(-0.02, 1.02)
ax.set_ylim(-0.02, 1.02)
ax.set_xticks([0, 0.5, 1])
ax.set_yticks([0, 0.5, 1])
ax.tick_params(labelsize=S.FONT_TICK - 1)
[s.set_visible(False) for s in (ax.spines["top"], ax.spines["right"])]
fig.text(
    _fx(_x0 + _side + 7),
    0.90,
    "Leave-one-batch-out\nblood APA classifier",
    fontsize=S.FONT_ANNOT,
    va="center",
    color="0.3",
)
fig.text(
    _fx(_x0 + _side + 7),
    0.55,
    f"pooled AUC = {A:.2f}",
    fontsize=S.FONT_LABEL,
    va="center",
    fontweight="bold",
)
fig.text(
    _fx(_x0 + _side + 7),
    0.30,
    "95% CI 0.55-0.82\n(excludes chance)",
    fontsize=S.FONT_ANNOT,
    va="center",
)
S.save_and_show(fig, "fig5d_classifier__eighth", figure="Fig5")
```

### Eighth-height ROC that fills the panel width

Alternative to the square-plot eighth version: the ROC spans the full panel width (stretched, so
the diagonal is shallow) with the AUC callout inside. Saved as `fig5d_classifier__eighth_wide`.

```python
# --- Eighth-height ROC that fills the panel width (stretched; AUC callout inside) ---
# Alternative to the square-plot eighth version (fig5d_classifier__eighth). Reuses fpr, tpr, A.
fig, ax = plt.subplots(figsize=(S.HALF_W_IN, S.EIGHTH_H_IN))
ax.plot(fpr, tpr, color="#2c7fb8", lw=1.5)
ax.fill_between(fpr, tpr, 0, color="#2c7fb8", alpha=0.10)
ax.plot([0, 1], [0, 1], ls="--", c="grey", lw=0.7)
ax.text(
    0.50,
    0.18,
    f"pooled AUC = {A:.2f}\n95% CI 0.55-0.82 (excludes chance)",
    fontsize=S.FONT_ANNOT - 0.5,
)
ax.set_xlabel("False positive rate", fontsize=S.FONT_TICK)
ax.set_ylabel("True positive rate", fontsize=S.FONT_TICK)
ax.set_xlim(-0.02, 1.02)
ax.set_ylim(-0.02, 1.02)
ax.set_xticks([0, 0.5, 1])
ax.set_yticks([0, 0.5, 1])
ax.tick_params(labelsize=S.FONT_TICK - 1)
[s.set_visible(False) for s in (ax.spines["top"], ax.spines["right"])]
fig.subplots_adjust(left=0.13, right=0.97, bottom=0.30, top=0.94)
S.save_and_show(fig, "fig5d_classifier__eighth_wide", figure="Fig5")
```
